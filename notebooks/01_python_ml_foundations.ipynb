## 2. Dataset Creation

We'll create a fictional e-commerce dataset with the following fields:

- `customer_id`: Unique identifier (int)
- `age`: Customer age (int)
- `gender`: M/F (str)
- `income`: Annual income in dollars (float)
- `products_viewed`: Number of products viewed (int)
- `products_purchased`: Number of products purchased (int)
- `average_order_value`: Average $ per order (float)
- `number_of_orders`: Total orders placed (int)
- `customer_rating`: Satisfaction rating 1-5 (float)

We'll start with ~50 customers to keep it manageable.

````python
# Generate synthetic customer data
def generate_customer_data(n_customers=50):
    """
    Generate synthetic e-commerce customer data.
    
    Parameters:
    -----------
    n_customers : int
        Number of customers to generate
        
    Returns:
    --------
    list of dict
        List where each element is a dictionary representing a customer
    """
    customers = []
    
    for i in range(1, n_customers + 1):
        # Generate realistic data with some correlations
        age = random.randint(18, 75)
        gender = random.choice(['M', 'F'])
        
        # Income correlates somewhat with age
        base_income = 30000 + (age - 18) * 1000
        income = base_income + random.uniform(-15000, 25000)
        income = max(20000, income)  # Minimum income
        
        # Viewing and purchasing behavior
        products_viewed = random.randint(5, 200)
        # Purchase rate varies (some browsers, some buyers)
        purchase_rate = random.uniform(0.01, 0.3)
        products_purchased = int(products_viewed * purchase_rate)
        
        # Order behavior
        if products_purchased > 0:
            number_of_orders = random.randint(1, max(1, products_purchased // 2))
            average_order_value = random.uniform(20, 500)
        else:
            number_of_orders = 0
            average_order_value = 0
        
        # Rating correlates with purchase behavior
        if number_of_orders > 0:
            base_rating = 3.0 + (purchase_rate * 5)
            customer_rating = min(5.0, max(1.0, base_rating + random.uniform(-1, 1)))
        else:
            customer_rating = random.uniform(1.0, 5.0)
        
        # Create customer dictionary
        customer = {
            'customer_id': i,
            'age': age,
            'gender': gender,
            'income': round(income, 2),
            'products_viewed': products_viewed,
            'products_purchased': products_purchased,
            'average_order_value': round(average_order_value, 2),
            'number_of_orders': number_of_orders,
            'customer_rating': round(customer_rating, 2)
        }
        
        customers.append(customer)
    
    return customers

# Generate the dataset
customers = generate_customer_data(50)

print(f"✓ Generated {len(customers)} customers")
print(f"✓ First customer: {customers[0]}")
````

## 3. Python Data Structures

Before analyzing data, let's understand the data structures we’re using. Each has unique properties that make it useful for different tasks in ML.

### Why Data Structures Matter for ML

ML requires you to store, access, and manipulate data efficiently. The choice of data structure affects:
- **Readability** of your code
- **Performance** of your operations
- **Ease of writing** transformations
- **Memory usage** of your programs

Python's main built-in data structures are:

1. **Lists** — ordered, mutable collections
2. **Tuples** — ordered, immutable collections
3. **Dictionaries** — key-value mappings
4. **Sets** — unordered collections of unique items

Let’s explore each one and see how they appear in ML workflows.

### 3.1 Lists

**Properties**:
- ✅ **Ordered**: Items have a fixed position
- ✅ **Mutable**: You can change, add, or remove items
- ✅ **Allows duplicates**: The same value can appear multiple times
- ✅ **Mixed types**: Can store different data types

**Use in ML**:
- Store a sequence of feature values for one sample
- Store rows of data
- Collect labels or predictions

**Common operations**: index, slice, append, extend, pop, sort, list comprehensions

````python
# Example: Working with lists
ages = [customer['age'] for customer in customers]

# Indexing and slicing
print("List Operations:")
print(f"All ages: {ages}")
print(f"First 5 ages: {ages[:5]}")
print(f"Last 5 ages: {ages[-5:]}")
print(f"Age at index 10: {ages[10]}")
print(f"Number of ages: {len(ages)}")
print(f"Youngest customer: {min(ages)}")
print(f"Oldest customer: {max(ages)}")

# Modifying a list
ages_copy = ages.copy()
ages_copy.append(30)
print(f"\nAfter append(30): {len(ages_copy)} ages")
ages_copy.pop()
print(f"After pop(): {len(ages_copy)} ages")
ages_copy.sort()
print(f"First 5 sorted ages: {ages_copy[:5]}")

# List of mixed feature types from one customer
first_customer_features = [customers[0]['age'], customers[0]['gender'], customers[0]['income']]
print(f"\nMixed feature list: {first_customer_features}")
print(f"Type of each item: {[type(x).__name__ for x in first_customer_features]}")
````

### 3.2 Tuples

**Properties**:
- ✅ **Ordered**: Items have a fixed position
- ✅ **Immutable**: Cannot be changed after creation
- ✅ **Allows duplicates**
- ✅ **Hashable**: Can be used as dictionary keys or set members

**Use in ML**:
- Store a fixed-size feature vector: `(age, income, rating)`
- Return multiple values from a function
- Use as keys when grouping by multiple fields
- Protect data from accidental modification

**Common operations**: index, slice, unpack, concatenate

---

### 3.3 Dictionaries

**Properties**:
- ✅ **Key-value pairs**: Access values by name
- ✅ **Mutable**: You can add, change, or remove items
- ✅ **No fixed order** (Python 3.7+ preserves insertion order)
- ✅ **Keys must be unique and hashable**

**Use in ML**:
- Store one row of structured data: `{'age': 25, 'income': 50000}`
- Map customer IDs to records
- Count class occurrences or word frequencies
- Store model hyperparameters: `{'learning_rate': 0.01, 'epochs': 100}`

**Common operations**: get, set, keys, values, items, update, pop

---

### 3.4 Sets

**Properties**:
- ✅ **Unordered**: No index or fixed position
- ✅ **Unique elements**: Duplicates are automatically removed
- ✅ **Mutable**: Can add or remove items
- ✅ **Fast membership testing**: `x in my_set` is O(1)

**Use in ML**:
- Find unique categories or labels
- Remove duplicate customer IDs
- Compute overlap between two groups
- Check if a feature value exists in a vocabulary

**Common operations**: add, remove, union, intersection, difference

---

### 3.5 Quick Comparison

| Structure | Ordered | Mutable | Duplicates | Best For |
|-----------|---------|---------|------------|----------|
| List | Yes | Yes | Yes | Sequences of data |
| Tuple | Yes | No | Yes | Fixed records, keys |
| Dictionary | No* | Yes | No (keys) | Named fields, mappings |
| Set | No | Yes | No | Unique items, fast lookup |

\* Dictionaries preserve insertion order in Python 3.7+

````python
# Helper: add age_bracket to each customer (used for grouping below)
def _get_age_bracket(age):
    """Categorize age into brackets."""
    if age < 26:
        return '18-25'
    elif age < 36:
        return '26-35'
    elif age < 46:
        return '36-45'
    else:
        return '46+'

for customer in customers:
    customer['age_bracket'] = _get_age_bracket(customer['age'])

# ===== TUPLES =====
# Store a fixed feature vector for one customer
first_customer = customers[0]
feature_tuple = (first_customer['age'], first_customer['income'], first_customer['customer_rating'])
print("Tuple example (feature vector):", feature_tuple)
print(f"Type: {type(feature_tuple)}")
print(f"Age from tuple: {feature_tuple[0]}")

# Tuples as keys: group customers by (gender, age_bracket)
for customer in customers:
    customer['tuple_key'] = (customer['gender'], customer['age_bracket'])

groups_by_tuple = {}
for customer in customers:
    key = customer['tuple_key']
    if key not in groups_by_tuple:
        groups_by_tuple[key] = []
    groups_by_tuple[key].append(customer)

print(f"\nGroups by (gender, age_bracket): {len(groups_by_tuple)} unique combinations")
for key, group in list(groups_by_tuple.items())[:3]:
    print(f"  {key}: {len(group)} customers")

# Tuple unpacking
age, income, rating = feature_tuple
print(f"\nUnpacked: age={age}, income={income}, rating={rating}")

# ===== DICTIONARIES =====
# Map customer_id to customer record
customer_by_id = {}
for customer in customers:
    customer_by_id[customer['customer_id']] = customer

print("\nDictionary lookup by customer_id:")
print(f"Customer 10: {customer_by_id[10]}")

# Use .get() to safely access missing keys
print(f"Customer 999: {customer_by_id.get(999, 'Not found')}")

# Count customers by gender using a dictionary
gender_counts = {}
for customer in customers:
    gender = customer['gender']
    gender_counts[gender] = gender_counts.get(gender, 0) + 1

print(f"\nGender counts: {gender_counts}")

# Count products viewed in each age bracket using .get()
views_by_bracket = {}
for customer in customers:
    bracket = customer['age_bracket']
    views_by_bracket[bracket] = views_by_bracket.get(bracket, 0) + customer['products_viewed']

print(f"\nTotal views by age bracket: {views_by_bracket}")

# ===== SETS =====
# Find unique genders
unique_genders = set()
for customer in customers:
    unique_genders.add(customer['gender'])

print(f"\nUnique genders: {unique_genders}")

# Find unique age brackets
unique_brackets = set(c['age_bracket'] for c in customers)
print(f"Unique age brackets: {unique_brackets}")

# Check if a value exists in a set
print(f"'26-35' in brackets: {'26-35' in unique_brackets}")

# Compare two customer groups using sets
high_income_ids = {c['customer_id'] for c in customers if c['income'] > 60000}
active_ids = {c['customer_id'] for c in customers if c['number_of_orders'] >= 3}

print(f"\nHigh income customers: {len(high_income_ids)}")
print(f"Active customers: {len(active_ids)}")
print(f"Customers who are both: {len(high_income_ids & active_ids)}")
print(f"Customers who are either: {len(high_income_ids | active_ids)}")
print(f"High income but not active: {len(high_income_ids - active_ids)}")

# Remove duplicates from a list of customer IDs
ids_with_duplicates = [1, 2, 2, 3, 3, 3, 4]
unique_ids = list(set(ids_with_duplicates))
print(f"\nRemoved duplicates: {unique_ids}")
````

## 4. Implementing Statistics Functions

Now let's implement basic statistical functions from scratch using pure Python.

**Why from scratch?** To understand what libraries like NumPy are doing for us.

````python
def calculate_mean(values):
    """Calculate the mean (average) of a list of numbers."""
    if not values:
        return None
    return sum(values) / len(values)

def calculate_median(values):
    """Calculate the median (middle value) of a list of numbers."""
    if not values:
        return None
    sorted_values = sorted(values)
    n = len(sorted_values)
    mid = n // 2
    
    if n % 2 == 0:
        # Even number of values: average of two middle values
        return (sorted_values[mid - 1] + sorted_values[mid]) / 2
    else:
        # Odd number of values: middle value
        return sorted_values[mid]

def calculate_mode(values):
    """Calculate the mode (most common value) of a list."""
    if not values:
        return None
    
    # Count occurrences
    counts = {}
    for value in values:
        counts[value] = counts.get(value, 0) + 1
    
    # Find maximum count
    max_count = max(counts.values())
    
    # Find all values with maximum count
    modes = [value for value, count in counts.items() if count == max_count]
    
    return modes[0] if len(modes) == 1 else modes

def calculate_variance(values):
    """Calculate the variance of a list of numbers."""
    if not values or len(values) < 2:
        return None
    
    mean = calculate_mean(values)
    squared_diffs = [(x - mean) ** 2 for x in values]
    return sum(squared_diffs) / (len(values) - 1)  # Sample variance

def calculate_std(values):
    """Calculate the standard deviation of a list of numbers."""
    variance = calculate_variance(values)
    return math.sqrt(variance) if variance is not None else None

# Test our functions
ages = [customer['age'] for customer in customers]

print("Statistical Functions (implemented from scratch):")
print(f"Mean age: {calculate_mean(ages):.2f}")
print(f"Median age: {calculate_median(ages):.2f}")
print(f"Mode age: {calculate_mode(ages)}")
print(f"Variance: {calculate_variance(ages):.2f}")
print(f"Standard deviation: {calculate_std(ages):.2f}")
print(f"Min age: {min(ages)}")
print(f"Max age: {max(ages)}")
print(f"Range: {max(ages) - min(ages)}")
````

## 5. Data Filtering and Selection

Learn to filter data based on conditions - a fundamental operation in data analysis.

````python
# Filter customers by age range
def filter_by_age(customers, min_age, max_age):
    """Return customers within specified age range."""
    return [c for c in customers if min_age <= c['age'] <= max_age]

# Filter by income threshold
def filter_by_income(customers, min_income):
    """Return customers with income above threshold."""
    return [c for c in customers if c['income'] >= min_income]

# Filter high-value customers
def filter_high_value(customers, min_order_value=100):
    """Return customers with high average order value."""
    return [c for c in customers if c['average_order_value'] >= min_order_value]

# Filter active customers
def filter_active(customers, min_orders=5):
    """Return customers with at least min_orders."""
    return [c for c in customers if c['number_of_orders'] >= min_orders]

# Test filtering
young_customers = filter_by_age(customers, 18, 30)
high_income = filter_by_income(customers, 60000)
high_value = filter_high_value(customers, 150)
active = filter_active(customers, 3)

print("Filtering Results:")
print(f"Total customers: {len(customers)}")
print(f"Young customers (18-30): {len(young_customers)}")
print(f"High income (>$60k): {len(high_income)}")
print(f"High value (AOV>$150): {len(high_value)}")
print(f"Active (3+ orders): {len(active)}")

# Combine filters
target_segment = [c for c in customers 
                  if 25 <= c['age'] <= 40 
                  and c['income'] >= 50000 
                  and c['number_of_orders'] >= 2]

print(f"\\nTarget segment (age 25-40, income>$50k, 2+ orders): {len(target_segment)}")
````

## 6. Data Grouping and Aggregation

Group customers by categories and calculate aggregate statistics.

````python
def group_by_field(customers, field):
    """
    Group customers by a specific field.
    
    Returns a dictionary where keys are field values and values are lists of customers.
    """
    groups = {}
    for customer in customers:
        key = customer[field]
        if key not in groups:
            groups[key] = []
        groups[key].append(customer)
    return groups

# Group by gender
by_gender = group_by_field(customers, 'gender')

print("Group by Gender:")
for gender, group in by_gender.items():
    incomes = [c['income'] for c in group]
    ages = [c['age'] for c in group]
    print(f"\\n{gender}:")
    print(f"  Count: {len(group)}")
    print(f"  Avg Age: {calculate_mean(ages):.1f}")
    print(f"  Avg Income: ${calculate_mean(incomes):,.2f}")

# Group by age bracket
def get_age_bracket(age):
    """Categorize age into brackets."""
    if age < 26:
        return '18-25'
    elif age < 36:
        return '26-35'
    elif age < 46:
        return '36-45'
    else:
        return '46+'

# Add age bracket to each customer (temporarily)
for customer in customers:
    customer['age_bracket'] = get_age_bracket(customer['age'])

by_age_bracket = group_by_field(customers, 'age_bracket')

print("\\n" + "="*60)
print("Group by Age Bracket:")
for bracket in ['18-25', '26-35', '36-45', '46+']:
    if bracket in by_age_bracket:
        group = by_age_bracket[bracket]
        orders = [c['number_of_orders'] for c in group]
        ratings = [c['customer_rating'] for c in group]
        print(f"\\n{bracket}:")
        print(f"  Count: {len(group)}")
        print(f"  Avg Orders: {calculate_mean(orders):.2f}")
        print(f"  Avg Rating: {calculate_mean(ratings):.2f}")
````

## 7. Comprehensive Analysis Report

Let's create a complete analysis report combining everything we've learned.

````python
def generate_customer_report(customers):
    """
    Generate a comprehensive customer analysis report.
    
    Returns a dictionary with key metrics and insights.
    """
    # Extract all numeric fields
    ages = [c['age'] for c in customers]
    incomes = [c['income'] for c in customers]
    products_viewed = [c['products_viewed'] for c in customers]
    products_purchased = [c['products_purchased'] for c in customers]
    avg_order_values = [c['average_order_value'] for c in customers if c['average_order_value'] > 0]
    num_orders = [c['number_of_orders'] for c in customers]
    ratings = [c['customer_rating'] for c in customers]
    
    # Calculate conversion rate
    total_viewed = sum(products_viewed)
    total_purchased = sum(products_purchased)
    conversion_rate = (total_purchased / total_viewed * 100) if total_viewed > 0 else 0
    
    # Calculate total revenue
    total_revenue = sum([c['average_order_value'] * c['number_of_orders'] for c in customers])
    
    # Identify high-value customers (top 20% by revenue)
    customer_revenues = [(c['customer_id'], c['average_order_value'] * c['number_of_orders']) 
                         for c in customers]
    customer_revenues.sort(key=lambda x: x[1], reverse=True)
    top_20_percent = int(len(customers) * 0.2)
    high_value_customers = customer_revenues[:top_20_percent]
    
    report = {
        'total_customers': len(customers),
        'demographics': {
            'avg_age': calculate_mean(ages),
            'median_age': calculate_median(ages),
            'age_range': (min(ages), max(ages)),
            'avg_income': calculate_mean(incomes),
            'median_income': calculate_median(incomes)
        },
        'behavior': {
            'avg_products_viewed': calculate_mean(products_viewed),
            'avg_products_purchased': calculate_mean(products_purchased),
            'conversion_rate': conversion_rate,
            'avg_orders_per_customer': calculate_mean(num_orders)
        },
        'revenue': {
            'total_revenue': total_revenue,
            'avg_order_value': calculate_mean(avg_order_values) if avg_order_values else 0,
            'revenue_per_customer': total_revenue / len(customers)
        },
        'satisfaction': {
            'avg_rating': calculate_mean(ratings),
            'median_rating': calculate_median(ratings)
        },
        'segments': {
            'high_value_count': len(high_value_customers),
            'high_value_revenue': sum([rev for _, rev in high_value_customers]),
            'high_value_percentage': (sum([rev for _, rev in high_value_customers]) / total_revenue * 100) if total_revenue > 0 else 0
        }
    }
    
    return report

# Generate the report
report = generate_customer_report(customers)

# Display the report
print("="*70)
print("CUSTOMER ANALYSIS REPORT")
print("="*70)

print(f"\\n📊 OVERVIEW")
print(f"Total Customers: {report['total_customers']}")

print(f"\\n👥 DEMOGRAPHICS")
print(f"Average Age: {report['demographics']['avg_age']:.1f} years")
print(f"Median Age: {report['demographics']['median_age']:.1f} years")
print(f"Age Range: {report['demographics']['age_range'][0]}-{report['demographics']['age_range'][1]} years")
print(f"Average Income: ${report['demographics']['avg_income']:,.2f}")
print(f"Median Income: ${report['demographics']['median_income']:,.2f}")

print(f"\\n🛒 BEHAVIOR")
print(f"Avg Products Viewed: {report['behavior']['avg_products_viewed']:.1f}")
print(f"Avg Products Purchased: {report['behavior']['avg_products_purchased']:.1f}")
print(f"Conversion Rate: {report['behavior']['conversion_rate']:.2f}%")
print(f"Avg Orders per Customer: {report['behavior']['avg_orders_per_customer']:.2f}")

print(f"\\n💰 REVENUE")
print(f"Total Revenue: ${report['revenue']['total_revenue']:,.2f}")
print(f"Average Order Value: ${report['revenue']['avg_order_value']:.2f}")
print(f"Revenue per Customer: ${report['revenue']['revenue_per_customer']:.2f}")

print(f"\\n⭐ SATISFACTION")
print(f"Average Rating: {report['satisfaction']['avg_rating']:.2f}/5.00")
print(f"Median Rating: {report['satisfaction']['median_rating']:.2f}/5.00")

print(f"\\n🎯 HIGH-VALUE SEGMENT")
print(f"High-Value Customers: {report['segments']['high_value_count']} ({report['segments']['high_value_count']/report['total_customers']*100:.0f}%)")
print(f"High-Value Revenue: ${report['segments']['high_value_revenue']:,.2f}")
print(f"% of Total Revenue: {report['segments']['high_value_percentage']:.1f}%")

print("\\n" + "="*70)
````
